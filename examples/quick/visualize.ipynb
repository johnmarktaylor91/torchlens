# Visualize a model graph

Draw the graph to see the layer labels before extracting or intervening. Rendering needs the Graphviz application (`dot` on your PATH) as well as `pip install torchlens[viz]`.

```python
import torch
from torch import nn

import torchlens as tl


def tiny_model(seed: int = 0) -> nn.Module:
    """A deterministic three-layer MLP."""

    torch.manual_seed(seed)
    return nn.Sequential(
        nn.Linear(4, 8),
        nn.ReLU(),
        nn.Linear(8, 2),
    )


torch.manual_seed(0)
model = tiny_model(seed=7).eval()
x = torch.randn(2, 4)
```

Trace the model and draw it; in a notebook the graph displays inline.

```python
log = tl.trace(model, x)
log.draw()
```

To write a file instead, pass a path: `log.draw("model_graph.svg")`. The node labels (`linear_1_1`, `relu_1_2`, `linear_2_3`) are the names `pluck`, `extract` and the intervention selectors take.
