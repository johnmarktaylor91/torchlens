# Save and load a `.tlspec`

`Trace.save` writes a trace to a `.tlspec` folder, activations included; `tl.load` reads it back.

```python
from pathlib import Path
from tempfile import TemporaryDirectory

import torch
from torch import nn

import torchlens as tl


def tiny_model(seed: int = 0) -> nn.Module:
    """A deterministic three-layer MLP."""

    torch.manual_seed(seed)
    return nn.Sequential(
        nn.Linear(4, 8),
        nn.ReLU(),
        nn.Linear(8, 2),
    )


torch.manual_seed(0)
model = tiny_model(seed=7).eval()
x = torch.randn(2, 4)
```

```python
with TemporaryDirectory() as tmpdir:
    path = Path(tmpdir) / "tiny_trace.tlspec"
    log = tl.trace(model, x, save="all")
    log.save(path, overwrite=True)
    loaded = tl.load(path)
    print(path.name)
    print(type(loaded).__name__)
    print(torch.allclose(log["relu_1_2"].out, loaded["relu_1_2"].out))
```

Pass `include_outs=False` to save the structure without the activations.
