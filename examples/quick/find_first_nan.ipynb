# Find the first NaN or Inf

When a model emits NaNs or infinities, TorchLens tells you which operation produced the first non-finite tensor.

```python
import torch
from torch import nn

import torchlens as tl


class NonFiniteFixture(nn.Module):
    """Tiny model that produces a NaN."""

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        """Divide zero by zero."""

        hidden = x + 1.0
        zero = hidden - hidden
        return zero / zero


torch.manual_seed(0)
model = NonFiniteFixture().eval()
x = torch.ones(1, 2)
```

Log the forward pass, then ask for the first non-finite layer.

```python
log = tl.trace(model, x)
print(log.first_nonfinite())
```

The result is text you can print or log; nothing raises.
