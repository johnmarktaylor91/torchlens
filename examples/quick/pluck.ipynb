# Pluck your first activation

`tl.pluck` returns one layer's output as a tensor, with no trace to keep.

```python
import torch
from torch import nn

import torchlens as tl


def tiny_model(seed: int = 0) -> nn.Module:
    """A deterministic three-layer MLP."""

    torch.manual_seed(seed)
    return nn.Sequential(
        nn.Linear(4, 8),
        nn.ReLU(),
        nn.Linear(8, 2),
    )


torch.manual_seed(0)
model = tiny_model(seed=7).eval()
x = torch.randn(2, 4)
```

```python
layer = "relu_1_2"
out = tl.pluck(model, x, layer)
out
```

The result is a normal PyTorch tensor, so you can immediately use `.shape`, `.mean()`, or any downstream tensor code.

```python
print(f"{layer}: shape={tuple(out.shape)}, mean={out.mean().item():.4f}")
```
